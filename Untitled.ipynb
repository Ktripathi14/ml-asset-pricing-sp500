```python
import numpy as np
import pandas as pd

PERIODS_PER_YEAR = 12
COSTS_BPS = [5, 10]
N_DECILES = 10

def longshort_with_turnover(df, pred_col):
    out, prev_long, prev_short = [], None, None
    for date, g in df.sort_values("Date").groupby("Date"):
        g = g.dropna(subset=[pred_col, "y_true"]).copy()
        if len(g) < N_DECILES * 2:
            continue
        g["dec"] = pd.qcut(g[pred_col].rank(method="first"),
                           N_DECILES, labels=False, duplicates="drop")
        longs = set(g.loc[g["dec"] == g["dec"].max(), "Ticker"])
        shorts = set(g.loc[g["dec"] == g["dec"].min(), "Ticker"])
        gross = (g.loc[g["Ticker"].isin(longs), "y_true"].mean()
                 - g.loc[g["Ticker"].isin(shorts), "y_true"].mean())
        if prev_long is None:
            tl = ts = 1.0
        else:
            tl = 1 - len(longs & prev_long) / max(len(longs), 1)
            ts = 1 - len(shorts & prev_short) / max(len(shorts), 1)
        prev_long, prev_short = longs, shorts
        out.append({"gross": gross, "tl": tl, "ts": ts})
    return pd.DataFrame(out)

def net_sharpe_row(df, pred_col, label, panel):
    ls = longshort_with_turnover(df, pred_col)
    ann = np.sqrt(PERIODS_PER_YEAR)
    row = {"panel": panel, "model": label,
           "gross_sharpe": ls["gross"].mean() / ls["gross"].std() * ann,
           "avg_turnover": ls[["tl", "ts"]].mean().mean()}
    for bps in COSTS_BPS:
        net = ls["gross"] - (bps / 1e4) * 2 * (ls["tl"] + ls["ts"])
        row[f"net_sharpe_{bps}bps"] = net.mean() / net.std() * ann
    return row

panels = {"full": "results/Results full new",
          "balanced": "results/results balanced new"}

rows = []
for panel, folder in panels.items():
    trad = pd.read_parquet(f"{folder}/pred_traditional.parquet")
    for fs in ["A", "B"]:
        ml = pd.read_parquet(f"{folder}/pred_ml_{fs}.parquet")
        merged = trad.merge(ml.drop(columns=["y_true"]),
                            on=["Date", "Ticker"], how="inner")
        if fs == "A":  # traditional models once per panel
            for m in ["CAPM", "FF3", "FF5"]:
                rows.append(net_sharpe_row(merged, f"pred_{m}", m, panel))
        for m in ["LASSO", "RF", "SVM", "NN"]:
            rows.append(net_sharpe_row(merged, f"pred_{m}", f"{m}_{fs}", panel))

result = pd.DataFrame(rows).round(3)
result.to_csv("net_sharpe_turnover.csv", index=False)
print(result.to_string(index=False))
```
